# Water Temperature (Streaming Table)

```python
@dlt.table(name="02_silver.water_temperature", comment="Water temperature enriched with metadata")
@dlt.expect_or_drop("valid_ingestion_timestamp", "ingestion_timestamp IS NOT NULL")
@dlt.expect("valid_water_temperature_range", "water_temperature BETWEEN 24.00 AND 30.00")
def silver_water_temperature():
    return (
        dlt.read_stream("01_bronze.water_temperature")
            .join(dlt.read("02_silver.buoy_metadata"), on="buoy_id", how="left")
            .select(
                col("buoy_id"),
                col("ingestion_timestamp"),
                col("simulation_timestamp"),                
                col("name"),
                col("manufacturer"),
                col("model"),
                col("production_year"),
                col("type"),
                col("power_source"),
                col("temperature_c").alias("water_temperature")
            )
    )
```
